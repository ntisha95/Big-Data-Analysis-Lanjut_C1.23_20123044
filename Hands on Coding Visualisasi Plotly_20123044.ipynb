```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df.head()
```

```python
# Tambah kolom IsAlone
df['IsAlone'] = (df['SibSp'] + df['Parch'] == 0)

df.head()
```

```python
fig1 = px.histogram(
    df, x="Sex", color="Survived",
    barmode="group",
    title="Survival berdasarkan Gender"
)
fig1.show()
```

```python
fig2 = px.histogram(
    df, x="Pclass", color="Survived",
    barmode="group",
    title="Survival berdasarkan Pclass"
)
fig2.show()
```

```python
fig3 = px.histogram(
    df, x="Age", color="Survived",
    title="Distribusi Age terhadap Survival"
)
fig3.show()
```

```python
fig4 = px.histogram(
    df, x="IsAlone", color="Survived",
    barmode="group",
    title="Survival berdasarkan IsAlone"
)
fig4.show()
```

Insight:

1. Siapa yang paling berpeluang selamat?
- Perempuan memiliki peluang selamat lebih tinggi dibanding laki-laki
- Penumpang kelas 1 (Pclass = 1) memiliki tingkat survival tertinggi

2. Faktor yang paling berpengaruh:
- Gender (faktor paling dominan)
- Pclass (kelas tiket)
- Fare (harga tiket)
- Usia (anak-anak lebih diutamakan)

```python
df['Fare_group'] = pd.qcut(df['Fare'], 3, labels=['Low Fare', 'Medium Fare', 'High Fare'])
```

```python
df_filtered = df[df['Age'] < 30]
```

```python
fig5 = px.scatter(
    df_filtered,
    x="Age",
    y="Fare",
    color="Survived",
    facet_col="Fare_group",
    hover_data=["Sex", "Pclass"],
    title="Scatter Plot Age vs Fare (Age < 30)"
)

fig5.show()
```

Insight Scatter Plot:

- Penumpang muda (Age < 30) dengan fare tinggi cenderung lebih banyak yang selamat
- Penumpang dengan fare rendah memiliki tingkat survival lebih rendah
- Perempuan tetap dominan selamat di semua kategori
- Kelas atas (Pclass 1) lebih banyak muncul di fare tinggi dan survival tinggi

Fitur interaktif yang digunakan:
- Zooming: pengguna bisa memperbesar grafik
- Hovering: menampilkan detail (Sex, Pclass, dll)
- Filtering visual: melalui legend (klik Survived)
- Facet: membagi grafik berdasarkan Fare_group
